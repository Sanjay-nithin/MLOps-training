```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv('../../dataset/smart factory machinery data set.csv')
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

# Single Linear Regression


```python
x = df[['Rotational speed [rpm]']]
y = df['Process temperature [K]']

model = LinearRegression()
model.fit(x, y)

model.predict([[1408]])
```

# Multivariate Linear Regression

```python
x = df.drop(['UDI', 'Product ID', 'Process temperature [K]'], axis='columns')
x['Type'] = x['Type'].replace({'M':0, "L":1, "H":2})
y = df['Process temperature [K]']

x.head()
```

```python
model = LinearRegression()
model.fit(x, y)

model.predict([[0,298.1,1551,42.8,0,0,0,0,0,0,0]])
```

# Saving the Model

```python
import pickle as pkl
import joblib
```

```python
with open('multivariate_LR.pkl', 'wb') as file:
    pkl.dump(model, file)

joblib.dump(model, 'multivariate_LR_joblib.pkl')
```

# One Hot Encoding

```python
x = df.drop(['UDI', 'Product ID', 'Process temperature [K]'], axis='columns')
y = df['Process temperature [K]']

new_x = pd.get_dummies(x, columns=['Type'], dtype=int)
new_x = new_x.drop('Type_M', axis=1)
new_x.head()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(new_x, y, test_size=0.2)

model = LinearRegression()
model.fit(x_train, y_train)

model.predict(x_test)
```

# Logistic Regression

```python
y = df['Machine failure']

x_train, x_test, y_train, y_test = train_test_split(new_x, y, test_size=0.2)
model = LogisticRegression()
model.fit(x_train, y_train)

model.predict(x_test)
```
